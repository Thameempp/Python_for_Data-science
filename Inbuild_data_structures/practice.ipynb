### Practicing Python Data structure

```python
# finding dublicates in a list
# method 1
lst = [1,2,3,4,5,6,6,7,8,9,8]

duplicates = list(set([x for x in lst if lst.count(x)>1]))
print(duplicates)
```

```python
# method 2 
seen = set()  # this store visited number 
duplicates = set()

for x in lst:
    if x in seen: # while visiting number , if it already there , means: this visiting number is repeated
        duplicates.add(x) # so it storing in duplicates
    else:
        seen.add(x)

print(list(duplicates))
```

```python
# remove duplicates from the list
print(list(set(lst))) # set do not store duplicates
```

```python
# count frequency of each element in a list
freq = {}
for x in lst:
    freq[x] = freq.get(x, 0) + 1

print(freq)
```

```python
# find largest element in a list 
print(max(lst))
```

```python
# finding smallest element
print(min(lst))
```

```python
# reversing a list
arr = [1,2,3,4,5,6,7]
arr.reverse()
print(arr)
```

```python
# Sort a list in ascending order
arr = [10,3,4,11,2,7,12,5,1,6,9,8]
arr.sort()
print(arr)
```

```python
# Sort list in descending order
arr.sort(reverse=True)
print(arr)
```

```python
# Find second largest element in list
lst = [2,40,5,9,6,7,8,9,10]

unique = list(set(lst))
unique.sort() # this will remove duplicate and sort to smallest to largest

print(unique[-2])
```

```python
# Merge two lists
a = [1, 2, 3]
b = [4, 5, 6]

c = a + b
print(c)
```

```python
# Find common elements in two lists
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]

common = list(set(a) & set(b)) # & finds intersection

print(common)
```

```python
# Add elements to a set
s = {1, 2, 3}

elements = [7,8,9,20]
for x in elements:
    s.add(x)
s
```

```python
# Remove element from set safely
# removing one element at a time 
s.discard(1)
s
```

```python
# removing multiple element at a time 
elmnt = [2,7,9]
for x in elmnt:
    s.discard(x)

s
```

```python
# Create dictionary from two lists
name = ['John', 'Alice']
age = [19, 20]

d = dict(zip(name, age))
print(d)
```

```python
# Print keys and values from dictionary
for k, v in d.items():
    print(k, v)
```

```python
# Find key with maximum value in dictionary
data = {'a': 10, 'b': 25, 'c': 15}
```

```python
max_key = max(data, key=data.get) # max(): compares keys , key=data.get : tells Python to compare values instead of keys
max_key
```

```python
for k in data:              # keys
    print(k)

for v in data.values():    # values
    print(v)

for k, v in data.items():  # key-value pairs
    print(k, v)
```
